```python
from pyspark import SparkConf, SparkContext
from pyspark.sql import SQLContext
conf = SparkConf().set('spark.ui.port', '4050').setAppName("Housing").setMaster("local[2]")
sc = SparkContext.getOrCreate(conf=conf)
sqlContext = SQLContext(sc)
```

```python
housing_df = sqlContext.read.format('com.databricks.spark.csv').options(header='true', inferschema='true').load(r'Housing.csv')
housing_df.printSchema()
```

```python
housing_df.toPandas()   #Creates a Pandas dataframe out of data
```

```python
housing_df.dtypes
```

```python
num_col=[i[0] for i in housing_df.dtypes if i[1]=="int"]
num_col
cat_col=[i[0] for i in housing_df.dtypes if i[1]!="int"]
cat_col
```

```python
from pyspark.ml.feature import VectorAssembler,StringIndexer,OneHotEncoder
stages=[]
for i in cat_col:
  string=StringIndexer(inputCol=i,outputCol=i+'Index')#An encoder defaults frequency encoding
  encoder=OneHotEncoder(inputCols=[string.getOutputCol()],outputCols=[i+"classVec"]) #Converting encoder column to OneHotEncoder
  stages+=[string,encoder]
```

```python
assembler=[c+"classVec" for c in cat_col]+num_col

vector_assembler=VectorAssembler(inputCols=assembler,outputCol="features")
stages+=[vector_assembler]
```

```python
from pyspark.ml import Pipeline
pipeline=Pipeline(stages=stages)
pipelinemodel=pipeline.fit(housing_df)
df=pipelinemodel.transform(housing_df)
```

```python
df.show(5)
```

```python
select_col=['price','features']
df=df.select(select_col)
df.show(3)
```

```python
splits = df.randomSplit([0.7, 0.3])
train_df = splits[0]
test_df = splits[1]
#train_df,test_df=vhouse_df.randomSplit([0.7,0.3])
from pyspark.ml.regression import DecisionTreeRegressor
dt = DecisionTreeRegressor(featuresCol ='features', labelCol = 'price')
dt_model = dt.fit(train_df)
dt_predictions = dt_model.transform(test_df)
from pyspark.ml.evaluation import RegressionEvaluator
dt_evaluator = RegressionEvaluator(labelCol="price", predictionCol="prediction", metricName="r2")
r2_score = dt_evaluator.evaluate(dt_predictions)
print("R2 Score on test data = %g" % r2_score)
```

```python
from pyspark.sql import SparkSession
spark=SparkSession.builder.appName('diabetes_data').getOrCreate()
```

```python
df1=spark.read.csv('Diabetes_prediction.csv',header=True,inferSchema=True)
df1.printSchema()
```

```python
df1.show(2)
```

```python
df1.describe().show()
```

```python
# prompt: select most columns except 2 as features using vector assembler

from pyspark.ml.feature import VectorAssembler
assembler = VectorAssembler(inputCols=[col for col in df1.columns if col not in ['Diagnosis']],outputCol='features')
assembled_df = assembler.transform(df1)
assembled_df.select('features').show(truncate=False)
df = assembled_df.select(['features', 'Diagnosis'])
df.show(3)
```

```python
df=df.withColumnRenamed("Diagnosis","label")
df.show(3)
```

```python
train,test=df.randomSplit([0.7,0.3])
print("Train Data Count: ",str(train.count()))
print("Test Data Count: ",str(test.count()))
```

```python
from pyspark.ml.classification import RandomForestClassifier
rf=RandomForestClassifier(featuresCol='features',labelCol="label")
rfModel=rf.fit(train)
predictions_rf=rfModel.transform(test)
predictions_rf.show(2)
from  pyspark.ml.evaluation import BinaryClassificationEvaluator
eval=BinaryClassificationEvaluator()
```

```python
print("Test Area UNder Roc: ",eval.evaluate(predictions_rf,{eval.metricName:"areaUnderROC"}))
```
